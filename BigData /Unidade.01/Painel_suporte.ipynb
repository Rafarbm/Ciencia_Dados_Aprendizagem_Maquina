```python
import pandas as pd

df = pd.read_csv("/content/chamados_suporte_ti.csv")

df.head()
```

```python
print("Dimensões:", df.shape)
print("Colunas:", df.columns.tolist())

df.info()
```

```python
df.describe()
```

```python
colunas = [
    "id_chamado",
    "categoria",
    "prioridade",
    "status",
    "tempo_resolucao_horas",
    "satisfacao",
    "reaberto"
]

analise = df[colunas].copy()

analise.head()
```

```python
resolvidos = analise[analise["status"] == "Resolvido"].copy()

resolvidos
```

```python
limites = {
    "Crítica": 4,
    "Alta": 8,
    "Média": 24,
    "Baixa": 48
}

resolvidos["limite_sla_horas"] = resolvidos["prioridade"].map(limites)
```

```python
resolvidos["dentro_sla"] = (
    resolvidos["tempo_resolucao_horas"]
    <= resolvidos["limite_sla_horas"]
)

resolvidos
```

```python
contagem = resolvidos["dentro_sla"].value_counts()

contagem
```

```python
percentual = (
    resolvidos["dentro_sla"]
    .value_counts(normalize=True)
    .mul(100)
    .round(1)
)

percentual
```

```python
media = (
    resolvidos
    .groupby("categoria")["tempo_resolucao_horas"]
    .mean()
    .sort_values(ascending=False)
    .round(2)
)

media
```

A análise mostrou que 11 dos 16 chamados resolvidos ficaram dentro do SLA, correspondendo a 68,8%, enquanto 5 chamados ficaram fora, representando 31,2%. A categoria Sistema lento apresentou o maior tempo médio de resolução, com 47,75 horas, influenciada pelo chamado de 120 horas. Recomenda-se investigar os chamados de Sistema lento e os casos que ultrapassaram o SLA para identificar as causas dos atrasos. Os chamados ainda não resolvidos não foram considerados no cálculo do tempo final porque ainda não possuem um tempo de resolução concluído.
