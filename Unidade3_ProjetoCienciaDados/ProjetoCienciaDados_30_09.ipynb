Criando o DataFrame das cidades


```python
import pandas as pd
cidades = pd.DataFrame(
    [
        ['Paraná', 'Londrina', 575377, 1356.00],
        ['São Paulo', 'São Carlos', 254484, 1508.00],
        ['Santa Catarina', 'Florianópolis', 508826, 1798.00],
        ['Paraná', 'Curitiba', 1963726, 2293.00],
        ['São Paulo', 'Campinas', 1223237, 1710.00]
    ],
    columns=['Estado', 'Cidade', 'Habitantes', 'Salário-Médio']
)

print("--- DataFrame Original ---")
display(cidades)
```

Funções Básicas

```python
print("\n--- Primeiras linhas (head) ---")
display(cidades.head(2))

print("\n--- Últimas linhas (tail) ---")
display(cidades.tail(2))

print("\n--- Dimensões (shape) ---")
print(cidades.shape)

print("\n--- Informações Gerais (info) ---")
cidades.info()

print("\n--- Estatísticas do Salário Médio ---")
display(cidades['Salário-Médio'].describe())
```

Usando os tipos de dados e datas

```python
cidades['Habitantes'] = cidades['Habitantes'].astype('float64')

# Adicionando coluna de data
cidades['Data'] = ['3-8-2021', '3-9-2021', '3-10-2021', '3-11-2021', '3-12-2021']
cidades['Data'] = pd.to_datetime(cidades['Data'], format='%d-%m-%Y')
```

```python
cidades['Data'] = cidades['Data'].dt.strftime('%d-%m-%Y')
```

Colunas

```python
cidades.rename(columns={'Habitantes': 'N Habitantes', 'Data': 'Data - Censo'}, inplace=True)
```

Adicionando e Removendo Colunas

```python
cidades.insert(2, 'Sigla', ['PR', 'SP', 'SC', 'PR', 'SP'])
cidades.insert(2, 'Sigla-Repetida', cidades['Sigla'])
```

```python
cidades = cidades.drop(['Sigla-Repetida'], axis=1)
```

Consultas

```python
print("\n--- Consulta: Salário maior que 1400.0 ---")
display(cidades.query('`Salário-Médio` > 1400.0'))
```

Ordenação

```python
cidades.sort_values(by='Salário-Médio', inplace=True)
print("\n--- Ordenado por Salário-Médio ---")
display(cidades)
```

```python
Combinando DataFrames
```

```python
cidades_salarios = pd.DataFrame([
    ['Paraná', 'Londrina', 1356.00],
    ['São Paulo', 'São Carlos', 1508.00],
    ['Santa Catarina', 'Florianópolis', 1798.00],
    ['Paraná', 'Curitiba', 2293.00],
    ['São Paulo', 'Campinas', 1710.00]
], columns=['Estado', 'Cidade', 'Salário-Médio'])

cidades_pop = pd.DataFrame([
    ['Londrina', 575377],
    ['São Carlos', 254484],
    ['Florianópolis', 508826],
    ['Curitiba', 1963726],
    ['Campinas', 1223237]
], columns=['Cidade', 'Habitantes'])

print("\n--- Teste de Merge (Inner Join) ---")
display(pd.merge(left=cidades_salarios, right=cidades_pop, on="Cidade"))
```

Salvando Arquivos

```python
cidades.to_csv("my_df.csv", index=False)
print("\nArquivo 'my_df.csv' salvo com sucesso no ambiente do Colab!")
```
